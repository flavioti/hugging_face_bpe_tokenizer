```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

# 1. Criar um dataset pequeno de teste
comentarios = [
    "eu gosto de programar em python",
    "python é uma linguagem de programação",
    "programar é muito divertido e aprender python é fácil",
]

with open("meu_dataset.txt", "w", encoding="utf-8") as f:
    for linha in comentarios:
        f.write(linha + "\n")

# 2. Inicializar o tokenizador BPE e o treinador
tokenizer = Tokenizer(BPE(unk_token="[UNK]"))
tokenizer.pre_tokenizer = Whitespace()
trainer = BpeTrainer(special_tokens=["[UNK]", "[PAD]", "[CLS]", "[SEP]"], vocab_size=50)

# 3. Treinar o tokenizador
tokenizer.train(["meu_dataset.txt"], trainer)

# =====================================================================
# 4. EXTRAIR E IMPRIMIR O CATÁLOGO EXATO (VOCABULÁRIO)
# =====================================================================
# O método get_vocab() retorna um dicionário do tipo {'palavra': ID}
vocabulo_completo = tokenizer.get_vocab()

# Vamos inverter e ordenar para exibir no formato: ID -> Token
vocabulo_ordenado = sorted(vocabulo_completo.items(), key=lambda item: item[1])

print("=== CATÁLOGO COMPLETO DO TOKENIZADOR (VOCABULÁRIO) ===")
for token, id_numerico in vocabulo_ordenado:
    # Substitui espaços visuais ocultos do BPE se houver, para facilitar leitura
    print(f"Gaveta {id_numerico:<3} -> {token}")
print("======================================================\n")

# 5. Testar a conversão na prática
texto_teste = "eu gosto de aprender"
output = tokenizer.encode(texto_teste)

print("=== TESTE DE TRADUÇÃO ===")
print("Texto Humano:", texto_teste)
print("Tokens (Pedaços):", output.tokens)
print("IDs que o LLM lê :", output.ids)

print("=== CENÁRIO 2: Caractere totalmente bizarro/novo ===")
# O caractere japonês '木' nunca foi visto e o alfabeto dele não foi mapeado.
teste_2 = "python 木"
output_2 = tokenizer.encode(teste_2)

print("Texto:", teste_2)
print("Tokens (Pedaços):", output_2.tokens)
print("IDs na gaveta   :", output_2.ids)
```
