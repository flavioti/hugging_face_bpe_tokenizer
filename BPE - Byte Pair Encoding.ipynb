```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace
```

# Criar um dataset pequeno de teste

```python
comentarios = [
    "ano passado eu morri, mas este ano eu não morro",
    "eu viajei no ano passado"
]

# Salvar o dataset em um arquivo de texto para o treinador ler
with open("meu_dataset.txt", "w", encoding="utf-8") as f:
    for linha in comentarios:
        f.write(linha + "\n")
```

# Inicializar um tokenizador BPE do zero

```python
tokenizer = Tokenizer(BPE(unk_token="[UNK]"))
tokenizer.pre_tokenizer = Whitespace() # Divide inicialmente por espaços
```

# Configurar o treinador (definindo um vocabulário máximo pequeno)



```python
trainer = BpeTrainer(special_tokens=["[UNK]", "[PAD]", "[CLS]", "[SEP]"], vocab_size=50)
```

# Treinar o tokenizador no seu dataset pequeno

```python
tokenizer.train(["meu_dataset.txt"], trainer)
```

# Testar a tokenização (Simulação)

```python
texto_teste = "vou viajar"
output = tokenizer.encode(texto_teste)
```

```python
print("Texto original:", texto_teste)
print("Tokens (Pedaços):", output.tokens)
print("IDs numéricos (o que o LLM lê):", output.ids)
```
